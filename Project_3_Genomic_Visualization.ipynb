```python
import matplotlib.pyplot as plt

def plot_gc_content(sequence, window_size=20):
    """
    Calculates GC content across a sliding window and plots it.
    """
    gc_values = []

    # The "Sliding Window" Logic
    for i in range(len(sequence) - window_size + 1):
        subseq = sequence[i:i + window_size]
        gc = (subseq.count('G') + subseq.count('C')) / len(subseq) * 100
        gc_values.append(gc)

    # Designing the Graph
    plt.figure(figsize=(12, 5))
    plt.plot(gc_values, color='#3498db', linewidth=1.5) # A nice professional blue
    plt.fill_between(range(len(gc_values)), gc_values, color='#3498db', alpha=0.1)

    plt.title('Genomic Visualization: GC Content Profile', fontsize=14)
    plt.xlabel('Position in Sequence (bp)', fontsize=12)
    plt.ylabel('GC %', fontsize=12)
    plt.axhline(y=50, color='gray', linestyle='--', alpha=0.5) # 50% marker
    plt.grid(True, alpha=0.2)
    plt.show()

# Let's test it with a "dummy" sequence first to make sure the graph works
test_sequence = "ATGC" * 50 # Creates a repeating pattern
plot_gc_content(test_sequence)
```

```python
# The Wuhan Spike Protein Sequence (Fragment)
wuhan_spike_dna = "ACTCAAAATTGCTGTTATTTATTACCAGATGATGTTTGTTTTTGAATTAGATGGTGATGTTAATGGGCACAAATTTTCTGTCAGTGGAGAG"

# Run the visualization on the real viral data!
print("Generating Visualization for SARS-CoV-2 Spike Protein Fragment...")
plot_gc_content(wuhan_spike_dna, window_size=10)
```
