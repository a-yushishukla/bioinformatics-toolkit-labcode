```python
from Bio import Align

# 1. Initialize the Aligner (The "Matching Engine")
aligner = Align.PairwiseAligner()
aligner.mode = 'global' # We want to align the full length of both sequences

# 2. Define two sequences that are slightly different
# (Imagine these are two bacterial strains from your TERI research)
seq1 = "GATTACA"
seq2 = "GTCGACG"

# 3. Perform the Alignment
alignments = aligner.align(seq1, seq2)

# 4. Display the best result
best_alignment = alignments[0]
print("--- Best Sequence Alignment ---")
print(best_alignment)
print(f"Alignment Score: {best_alignment.score}")
```

```python
from Bio import Align

# 1. Initialize the Aligner (The "Matching Engine")
aligner = Align.PairwiseAligner()
aligner.mode = 'global' # We want to align the full length of both sequences

# 2. Define two sequences that are slightly different
# (Imagine these are two bacterial strains from your TERI research)
seq1 = "GATTACA"
seq2 = "GATCA"

# 3. Perform the Alignment
alignments = aligner.align(seq1, seq2)

# 4. Display the best result
best_alignment = alignments[0]
print("--- Best Sequence Alignment ---")
print(best_alignment)
print(f"Alignment Score: {best_alignment.score}")
```

### Project 1: DNA Sequence Alignment Analysis

#### Scenario A: Substitution Analysis
* **Sequences:** GATTACA vs GTCGACG (7bp each)
* **Observation:** Identified 3 matches and 4 mismatches. No gaps were introduced.
* **Result:** Alignment Score: 3.0

#### Scenario B: Indel (Insertion/Deletion) Analysis
* **Sequences:** GATTACA (7bp) vs GATCA (5bp)
* **Observation:** The aligner introduced a 2-base gap (`--`) to maximize the score.
* **Biological Significance:** This demonstrates the algorithm's ability to detect deletions, which are common in viral evolution (like the deletions seen in different COVID-19 variants).
