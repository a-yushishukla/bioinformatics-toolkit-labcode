```python
from Bio.Phylo.TreeConstruction import DistanceMatrix, DistanceTreeConstructor
from Bio import Phylo
import matplotlib.pyplot as plt

# 1. Define our Variants and Genetic Distances
# (Distances based on shared mutations: Wuhan is the root)
names = ["Wuhan", "Delta", "Omicron"]

# The matrix is 'lower triangular' (Distance from self is 0)
matrix = [
    [0],                          # Wuhan
    [0.12, 0],                    # Delta (12% distance from Wuhan)
    [0.28, 0.24, 0]               # Omicron (Further from both)
]

dm = DistanceMatrix(names, matrix)

# 2. Build the Tree using UPGMA (Unweighted Pair Group Method)
constructor = DistanceTreeConstructor()
tree = constructor.upgma(dm)

# 3. Visualization
fig = plt.figure(figsize=(10, 6), dpi=100)
axes = fig.add_subplot(1, 1, 1)

# Draw the tree with basic styling
Phylo.draw(tree, axes=axes, do_show=False)
plt.title("Phylogenetic Relationship: SARS-CoV-2 Variants")

# 4. Save the result for your Portfolio
plt.savefig("variant_tree.png")
plt.show()

print("✅ Tree generated and saved as 'variant_tree.png'!")
```
