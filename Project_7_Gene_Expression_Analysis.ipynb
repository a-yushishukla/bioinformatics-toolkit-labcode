```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Create the Data (Gene Expression Levels)
# Values represent mRNA counts (higher = more active)
data = {
    'Healthy_Control_1': [1.2, 0.5, 5.2, 0.8],
    'Healthy_Control_2': [1.1, 0.4, 5.0, 0.9],
    'Treated_Patient_1': [4.5, 0.6, 0.2, 3.9],
    'Treated_Patient_2': [4.8, 0.5, 0.1, 4.1]
}

# Famous Genes:
# BRCA1 (Repair), TP53 (Guardian), MYC (Growth), EGFR (Signaling)
genes = ['BRCA1', 'TP53', 'MYC', 'EGFR']
df = pd.DataFrame(data, index=genes)

# 2. Plotting the Heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(df, annot=True, cmap='RdYlBu_r', center=2.0)

plt.title('Gene Expression Profile: Healthy vs. Treated')
plt.xlabel('Samples')
plt.ylabel('Critical Genes')

# 3. Save for the Portfolio
plt.savefig("expression_heatmap.png")
plt.show()

print("✅ Project 7: Heatmap saved as 'expression_heatmap.png'")
```
